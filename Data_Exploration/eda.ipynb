```python
Q1  = data["Fare"].quantile(0.25)
Q3 = data["Fare"].quantile(0.75)
IQR = Q3 - Q1
Upper_bound = Q3 + 1.5 * IQR
Lower_bound = Q1 - 1.5 * IQR

print(f"Upper bound: {Upper_bound}")
print(f"Lower bound {Lower_bound}")
```

```python
high_outliers = data["Fare"] > Upper_bound
low_outliers = data["Fare"] < Lower_bound
print(f"Upper bound {Upper_bound} and lower bound {Lower_bound}")
outliers = high_outliers | low_outliers
percentage = (outliers.sum()/len(data))*100
print(f"percentage {percentage}")
outliers_rows = data[outliers]
outliers_rows["Fare"].head()

```
